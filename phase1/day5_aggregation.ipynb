```python
import numpy as np
```

```python
data = np.array([60, 70, 80, 90, 100])
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
```

```python
matrix = np.array([
    [60, 70, 80],
    [90, 100, 110],
    [40, 50, 60]
])

print(np.mean(matrix))
print(np.mean(matrix, axis=0))
print(np.mean(matrix, axis=1))
```

```python
array = np.random.randint(1, 11, size= (4,5))
print(array)
print("Mean: ", np.mean(array, axis=1))
print("STD:", np.std(array, axis=0))
print("Index Nilai Maksimum: ", np.argmax(array, axis=1))

#cek manual

newarray = array[0, 0]
print(newarray)
```
