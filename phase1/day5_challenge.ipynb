```python
import numpy as np
```

```python
array = np.random.randint(1, 101, size = (5,4))

print("Mean", np.mean(array))
print("Median", np.median(array))
print("Std", np.std(array))

print("Mean by column", np.mean(array, axis=0))
print("Std by row", np.std(array, axis=1))
```

```python
scores = np.array([45, 78, 92, 61, 88, 73, 55, 96, 42, 84])

print("Mean:", np.mean(scores))
print("Median:", np.median(scores))

print("Apakah mean lebih besar dari median?", np.mean(scores) > np.median(scores))

print("Index tertinggi:", np.argmax(scores))
print("Index terendah:", np.argmin(scores))
```

```python
random = np.random.randint(1, 101, size=(3,4))

mean = np.mean(random)
print("Jumlah Elemen diatas mean:", len(random[random > mean]))
print("Elemen diatas mean:", random[random > mean])

print("Persen:", (len(random[random > mean]) / random.size) * 100, "%")
```
