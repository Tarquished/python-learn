```python
import numpy as np
```

```python
def normalize(matrix):
    mean = np.mean(matrix)
    std = np.std(matrix)
    normalized = (matrix - mean) / std
    return normalized

array = np.random.randint(1, 25, size=(4,4))

print(array)
print("mean sebelum normalized: ", np.mean(array))
print("std sebelum normalized: ", np.std(array))
print(normalize(array))
print("mean habis normalized:", np.mean(normalize(array)))
print("std habis normalized:", np.std(normalize(array)))
```
